## Loading PDFs

```python
from config import Config
Config.print_config()
```

```python
from pdf_reader import PDFReader

# 1. Crear el lector de PDFs
reader = PDFReader()

# 2. Leer TODOS los PDFs de la carpeta
documentos = reader.read_pdf_folder(str(Config.PDFS_DIR))
print(f"\nTotal documentos procesados: {len(documentos)}")
```

```python
# 3. Resumen de los documentos cargados (con metadata)
print(f"{'#':<3} {'Fuente':<38} {'Categoría':<20} {'Caracteres':>10}")
print("-" * 80)
for i, doc in enumerate(documentos, 1):
    m = doc["metadata"]
    print(f"{i:<3} {m['source'][:37]:<38} {m['category']:<20} {len(doc['text']):>10,}")
```
