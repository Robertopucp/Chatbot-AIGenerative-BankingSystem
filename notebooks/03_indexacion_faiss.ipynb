# 03 · Indexación en FAISS

Objetivo: indexar los **catálogos de productos** (TECHSTORE) en una base
vectorial FAISS persistente, y verificar que el índice se puede guardar y
recargar desde disco.

```python
import sys
from pathlib import Path

# Asegura que el notebook encuentre los módulos del proyecto (config, rag_service, ...)
sys.path.insert(0, str(Path.cwd().parent))
```

```python
from config import Config
from rag_service import RAGService

# 1. Crear el servicio RAG (embeddings de OPEN AI)
rag = RAGService()
Config.print_config()
```

```python
# 2. Indexinf 12 INDECOPI resolutions and store them in FAISS

num_chunks = rag.load_documents_from_pdfs()

print(f"\n✓ {num_chunks} chunks indexados en FAISS")
```

```python
# 3. Estadísticas del índice recién creado
print("Estadísticas del índice:")
for k, v in rag.get_stats().items():
    print(f"  {k}: {v}")
```

```python
# 4. Verificar persistencia: recargar el índice desde disco
rag2 = RAGService()
if rag2.load_index():
    print("\nÍndice recargado:")
    for k, v in rag2.get_stats().items():
        print(f"  {k}: {v}")
```
